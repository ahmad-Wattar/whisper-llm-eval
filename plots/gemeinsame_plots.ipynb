```python
import numpy as np
import matplotlib.pyplot as plt

models = ["GPT OSS 120B","Llama 3.1 8B","MedGemma 27B","SauerkrautLM 70B"]
embeddings = ["Jina_V3","GTE","BGE-M3","Aari1995","LlamaEmb"]

mean = np.array([
    [0.579245, 0.566712, 0.538573, 0.278504, 0.820213],
    [0.578342, 0.586652, 0.617068, 0.373000, 0.777759],
    [0.585380, 0.563973, 0.563238, 0.280483, 0.845996],
    [0.630696, 0.601717, 0.676842, 0.387460, 0.805437],
])

std = np.array([
    [0.047680, 0.070717, 0.058310, 0.045754, 0.068678],
    [0.074825, 0.070760, 0.089265, 0.100373, 0.107591],
    [0.082087, 0.063949, 0.073856, 0.058251, 0.123048],
    [0.053615, 0.050781, 0.061510, 0.051632, 0.068874],
])

fig, ax = plt.subplots(figsize=(10, 4.2))

# y-Positionen: pro Embedding ein Block, darin 4 Modelle
gap = 1.3
base = np.arange(len(embeddings)) * (len(models) + gap)
ypos = np.vstack([base + i for i in range(len(models))]).T  # (embeddings, models)

for i, m in enumerate(models):
    ax.errorbar(
        mean[i], ypos[:, i],
        xerr=std[i],
        fmt="o",
        capsize=3,
        label=m
    )

# Embedding-Labels in der Mitte jedes Blocks
ax.set_yticks(base + (len(models)-1)/2)
ax.set_yticklabels(embeddings)

ax.set_xlabel("L2 distance (lower is better)")
ax.set_title("Transcript–Summary L2 Distance (Mean ± Std)")
ax.legend(ncols=2, fontsize=8, frameon=False)
ax.grid(True, axis="x", linewidth=0.5)
plt.tight_layout()
plt.savefig("L2_Distance.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Data
# -----------------------------
models = ["Aari1995", "Jina_V3", "GTE", "BGE-M3", "LlamaEmb"]

cosine_full = np.array([60, 30, 50, 40, 80])
l2_full     = np.array([60, 30, 50, 40, 80])

cosine_filt = np.array([83.3, 33.3, 50.0, 33.3, 83.3])
l2_filt     = np.array([83.3, 33.3, 50.0, 50.0, 83.3])

x = np.arange(len(models))
width = 0.18

# -----------------------------
# Style (paper-friendly)
# -----------------------------
plt.rcParams.update({
    "font.size": 8,
    "axes.titlesize": 9,
    "axes.labelsize": 8,
    "xtick.labelsize": 7,
    "ytick.labelsize": 7,
    "legend.fontsize": 7,
})

# -----------------------------
# Plot
# -----------------------------
fig, ax = plt.subplots(figsize=(6, 3))  # kompakter!

# Balken
ax.bar(x - 1.5*width, cosine_full, width, label="Cos (10)")
ax.bar(x - 0.5*width, l2_full,     width, label="L2 (10)")

ax.bar(x + 0.5*width, cosine_filt, width, label="Cos (6)")
ax.bar(x + 1.5*width, l2_filt,     width, label="L2 (6)")

# Achsen
ax.set_xticks(x)
ax.set_xticklabels(models, rotation=30, ha="right")

ax.set_ylabel("Agreement (%)")

# Titel im Paper meist optional → lieber in Caption
# ax.set_title("Pairwise Trend Agreement")

ax.set_ylim(0, 100)

# Legende kompakt oben
ax.legend(
    ncols=2,
    frameon=False,
    loc="upper center",
    bbox_to_anchor=(0.5, 1.15)
)

# Optional: dezente horizontale Linien
ax.spines["top"].set_visible(False)
ax.grid(axis="x", visible=False)
ax.spines["right"].set_visible(False)
ax.grid(axis="y", linestyle="--", linewidth=0.5, alpha=0.5)

plt.tight_layout()

# Speichern (sehr wenig Padding)
plt.savefig(
    "pairwise_trend_Agreement.png",
    dpi=300,
    bbox_inches="tight",
    pad_inches=0.02
)

plt.show()
```
